```python
#!/usr/bin/env python3
"""
tgju_symbols.py
Build a {persian_name: code} dictionary of the instruments listed on tgju.org.

Example output:
    {"طلای 18 عیار": "geram18", "دلار": "price_dollar_rl", "بیت کوین": "crypto-bitcoin", ...}

`code` is tgju's internal slug — the same token used in the profile URL
    https://www.tgju.org/profile/<code>
and in tgju's JSON feeds.

Requires:  pip install requests beautifulsoup4
Run:       python tgju_symbols.py
"""

import json
import re
import sys
import requests
from bs4 import BeautifulSoup

# Pages that render server-side market tables. Edit / add freely.
PAGES = [
    "https://www.tgju.org/",           # dashboard: gold, coin, currency, energy
    "https://www.tgju.org/gold-chart",  # gold & precious metals
    "https://www.tgju.org/coin",        # coins
    "https://www.tgju.org/currency",    # fiat currencies
    "https://www.tgju.org/crypto",      # cryptocurrencies
]

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
        "(KHTML, like Gecko) Chrome/124.0 Safari/537.36"
    ),
    "Accept-Language": "fa-IR,fa;q=0.9,en;q=0.8",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
}

# tgju stores the instrument code on the <tr> in one of these attributes.
CODE_ATTRS = ("data-market-row", "data-market-nameslug", "data-market-p")

# Crypto rows use bilingual labels ("بیت کوین / Bitcoin"). When True, the
# trailing Latin half is trimmed -> "بیت کوین". Set False to keep the full label.
STRIP_LATIN_SUFFIX = True
_LATIN_SUFFIX_RE = re.compile(r"\s*/\s*[\x00-\x7F]+$")


def _clean_name(name):
    name = " ".join(name.split())
    if STRIP_LATIN_SUFFIX:
        name = _LATIN_SUFFIX_RE.sub("", name).strip() or name
    return name


def _extract_code(tr):
    """Return the instrument code for a <tr>, or None."""
    # 1) Preferred: the data-* attribute tgju puts on each market row.
    for attr in CODE_ATTRS:
        val = tr.get(attr)
        if val and val.strip():
            return val.strip()
    # 2) Fallback: the slug inside a /profile/<code> link.
    a = tr.select_one('a[href*="/profile/"]')
    if a and a.get("href"):
        tail = a["href"].split("/profile/")[-1]
        return tail.split("/")[0].split("?")[0].strip() or None
    return None


def _extract_name(tr):
    """Return the Persian label for a <tr>, or None."""
    th = tr.find("th")                            # gold / coin / currency tables
    if th:
        name = th.get_text(" ", strip=True)
        if name:
            return _clean_name(name)
    a = tr.select_one('a[href*="/profile/"]')     # crypto rows link to the profile
    if a:
        name = a.get_text(" ", strip=True)
        if name:
            return _clean_name(name)
    td = tr.find("td")                            # last resort: first data cell
    if td:
        name = td.get_text(" ", strip=True)
        if name:
            return _clean_name(name)
    return None


def _collect_from_html(html, symbols):
    """Parse one page's HTML and add {name: code} pairs to `symbols`."""
    soup = BeautifulSoup(html, "html.parser")
    found = 0
    for tr in soup.find_all("tr"):
        code = _extract_code(tr)
        name = _extract_name(tr)
        if code and name:
            symbols[name] = code
            found += 1
    return found


def fetch(url):
    resp = requests.get(url, headers=HEADERS, timeout=25)
    resp.raise_for_status()
    resp.encoding = resp.apparent_encoding or "utf-8"
    return resp.text


def get_symbols(pages=PAGES, verbose=True):
    """Scrape every page with plain requests and return {persian_name: code}."""
    symbols = {}
    for url in pages:
        try:
            html = fetch(url)
        except requests.RequestException as exc:
            if verbose:
                print(f"[skip] {url} -> {exc}", file=sys.stderr)
            continue
        found = _collect_from_html(html, symbols)
        if verbose:
            print(f"[ok]   {url} -> {found} rows", file=sys.stderr)
    return symbols


def get_symbols_selenium(pages=PAGES, wait=8, verbose=True):
    """
    Fallback that renders JavaScript before scraping. Use only if plain requests
    returns nothing (anti-bot) or a page's rows are injected by JS.
    Requires:  pip install selenium   + a matching browser driver (e.g. geckodriver).
    """
    import time
    from selenium import webdriver

    symbols = {}
    opts = webdriver.FirefoxOptions()
    opts.add_argument("--headless")
    with webdriver.Firefox(options=opts) as driver:
        for url in pages:
            driver.get(url)
            time.sleep(wait)                       # let the tables render
            found = _collect_from_html(driver.page_source, symbols)
            if verbose:
                print(f"[ok]   {url} -> {found} rows", file=sys.stderr)
    return symbols


def main():
    symbols = get_symbols()
    if not symbols:
        print("No symbols found — tgju may have served an anti-bot / JS page.\n"
              "Try the Selenium fallback:  get_symbols_selenium()", file=sys.stderr)
        sys.exit(1)
    print(symbols)
    # print(json.dumps(symbols, ensure_ascii=False, indent=2))
    # with open("tgju_symbols.json", "w", encoding="utf-8") as fh:
    #     json.dump(symbols, fh, ensure_ascii=False, indent=2)
    # print(f"\n{len(symbols)} symbols written to tgju_symbols.json", file=sys.stderr)


if __name__ == "__main__":
    main()
```

```python
def get_tsetmc_all_index():
    url = f"https://cdn.tsetmc.com/api/Index/GetIndexB1LastAll/All/1"
    
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
        "Referer": "https://www.tsetmc.com/",
        "Origin": "https://www.tsetmc.com",
        "Accept": "application/json",
    }
    
    r = requests.get(url, headers=headers, timeout=15)
    r.raise_for_status()
    payload = r.json()
    return payload["indexB1"]

def get_tgju(symbol:str='geram18'):
    url = f"https://api.tgju.org/v1/market/indicator/summary-table-data/{symbol}"
    r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
    return r.json()["data"]

def get_tsetmc_index(insCode:int=67130298613737946):
    url = f"https://cdn.tsetmc.com/api/Index/GetIndexB2History/{insCode}"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
        "Referer": "https://www.tsetmc.com/",
        "Origin": "https://www.tsetmc.com",
        "Accept": "application/json",
    }
    
    r = requests.get(url, headers=headers, timeout=15)
    r.raise_for_status()
    payload = r.json()
    return payload["indexB2"]
```

```python
dff = gold_df.merge(tidx_df, left_index=True, right_index=True, how='outer').ffill()
df=dff[['close_x','close_y']].dropna().rename(columns={"close_x": "gold", "close_y": "tbi"})
daily=df.reindex(
    pd.date_range(df.index.min(), df.index.max(), freq='D')
).ffill()

days = 365
gpct = f'g_pct_{days}d'
bpct = f'b_pct_{days}d'
df[gpct] = (daily['gold'] / daily['gold'].shift(days) - 1).reindex(df.index)
df[bpct] = (daily['tbi']  / daily['tbi' ].shift(days) - 1).reindex(df.index)
import jdatetime
df['j_date'] = df.index.map(lambda x: jdatetime.date.fromgregorian(date=x))
df
```
